```python
import pandas as pd

df = pd.read_csv("Resume.csv")
df.head()
```

```python
df = df[['Resume_str', 'Category']]
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
import re

def clean_text(text):
    text = text.lower()  
    text = re.sub(r'\W', ' ', text) 
    text = re.sub(r'\s+', ' ', text) 
    return text

df['cleaned_resume'] = df['Resume_str'].apply(clean_text)

df.head()
```
